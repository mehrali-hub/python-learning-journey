```python
import random 

# This generates a random number between 1 and 100
a = random.randint(1, 100)

# Initial guess
guess = int(input('Guess the number: '))

# The loop continues as long as the guess is wrong
while guess != a:
    if guess < a:
        print('Wrong guess! Guess higher.')
    else:
        print('Wrong! Guess lower.')
    
    # Crucial: Ask for the next guess inside the loop
    guess = int(input('Try again: '))

# This only runs once the loop finishes (when guess == a)
print('Correct guess! You won.')
```
